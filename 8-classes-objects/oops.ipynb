```python
class Person:
    def __init__(self, name:str, age:int):
        self.name = name
        self.age = age
    
    def can_vote(self):
        return self.age > 18
    
p1 = Person("Vineet", 21)

print(p1.can_vote())
print(p1.__dict__)
```

```python
class BankAccount:
    def __init__(self, owner: str, initial_balance: float) -> None:
        self.owner = owner
        self.balance = initial_balance

    def deposit(self, amount: float) -> float:
        self.balance += amount
        return self.balance
    
    def withdraw(self, amount: float) -> float:
        self.balance -= amount
        return self.balance
    
    def get_balance(self) -> float:
        return self.balance
    
my_account = BankAccount("Vineet", 1000)

print(my_account.get_balance())
print(my_account.deposit(2000))
print(my_account.withdraw(200.5))
```
