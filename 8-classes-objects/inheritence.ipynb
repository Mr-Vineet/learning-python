```python
class Pokemon:
    def __init__(self, category, name):
        self.category = category
        self.name = name
    
    def get_type(self):
        return self.category

    def get_name(self):
        return self.name
```

```python
class ElectricPokemon(Pokemon):
    def __init__(self, name, category, attack):
        super().__init__(category, name)
        self.attacks = attack
    
    def attack(self):
        return f"{str.capitalize(self.get_name())} attacked with {self.attacks}"
```

```python
pikachu = ElectricPokemon('Pikachu', 'electric', 'lightening bolt')

print(pikachu.get_name())
print(pikachu.get_type())
print(pikachu.attack())
```

```python
# multiple inheritence is when the class inherits more than one base class
class Animal:
    def __init__(self, name):
        self.name = name
    
    def says(self):
        return f'{self.name} says something'
    
class Pet:
    def __init__(self, owner):
        self.owner = owner
    
    def get_owner(self):
        return self.owner

class Dog(Pet, Animal):
    def __init__(self, owner, name, is_friendly):
        
        Pet.__init__(self, owner)
        Animal.__init__(self, name)
        self.is_friendly = is_friendly

    def says(self):
        return f"{self.name} says woof woof"

bruno = Dog("Vineet", "Bruno", True)

print(bruno.__dict__)
print(bruno.get_owner())
print(bruno.says())
bruno.is_friendly = False
print(bruno.is_friendly)
```
