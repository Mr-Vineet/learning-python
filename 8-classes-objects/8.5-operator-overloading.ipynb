```python
class TwoDimensional:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return TwoDimensional(self.x + other.x, self.y + other.y)
    
    def __repr__(self):
        return f"TwoDimensional{{ x={self.x}, y={self.y} }}"
    
c1 = TwoDimensional(2,3)
c2 = TwoDimensional(3,3)
print(c1 + c2)
```

```python
## class of complex number
class ComplexNumber:
    def __init__(self, real, img):
        self.real = real
        self.img = img
    
    def __add__(self, other):
        return ComplexNumber(self.real + other.real, self.img + other.img)
    
    def __sub__(self, other):
        return ComplexNumber(self.real - other.real, self.img - other.img)
    
    def __mul__(self, other):
        real_part = self.real * other.real - self.img * other.img
        img_part = self.img * other.real + self.real * other.img

        return ComplexNumber(real_part, img_part)
    
    def __truediv__(self, other):
        real_part = self.real * other.real + self.img * other.img
        img_part = img_part = self.img * other.real - self.real * other.img
        denominator = other.real ** 2 + other.img ** 2

        return ComplexNumber(real_part / denominator, img_part / denominator)

    def __eq__(self, other):
        return isinstance(other, ComplexNumber) and self.real == other.real and self.img == other.img 

    def __repr__(self):
        return f"ComplexNumber{{{self.real} + {self.img}i}}"

num1 = ComplexNumber(8, -3)
num2 = ComplexNumber(0, 2)

print(num1 + num2)
print(num1 - num2)
print(num1 * num2)
print(num1 / num2)
num1 == num1
```
