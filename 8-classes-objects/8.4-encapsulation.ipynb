```python
# Remember the rules aren't strict here it's just the intension you are showing (Access modifiers are conventions and not strict in terms of working of a program)

class Person:
    def __init__(self, name: str, age: int, nationality: str):
        self.name = name # public field
        self.__nationality = nationality # private field
        self._age = age # protected field

class Employee(Person):
    def __init__(self, name, age, nationality):
        super().__init__(name, age, nationality)
    
    def is_eligible_for_work(self):
        return self._age > 17

pika = Employee("Pika", 26, "Indian")

def get_name(person):
    return person.name

print(dir(pika))
print(get_name(pika))
pika._age = 10
print(pika._age)
print(pika.is_eligible_for_work())
```

```python
class Person:
    def __init__(self, name: str, age: int, nationality: str):
        self.name = name # public field
        self.__nationality = nationality # private field
        self.__age = age # private field

    def get_age(self):
        return self.__age
    
    def set_age(self, age):
        if(age > 0):
            self.__age = age
            return "age is updated"
        else:
         return "age can't be updated"
    
person = Person("A", 24, "Indian")
person.set_age(-10)
# person.get_age()
```

```python
# Magic methods

class Person:
    def __init__(self, name: str, age: int, nationality: str):
        self.name = name
        self.age = age
    
    def __str__(self):
        return f"Person{{name={self.name}, age={self.age}}}"
    

        
jadu = Person("jadu", 100, "marsey")
jadu2 = Person("jadu2", 100, "marsey")
dir(jadu)
```

```python
jadu.__str__()
str(jadu)
print(jadu + jadu2)
```
