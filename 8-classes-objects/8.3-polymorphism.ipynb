```python
from abc import ABC, abstractmethod
from math import *

class Shape(ABC):
    @abstractmethod
    def area(self) -> float:
        pass
    
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return pi * (self.radius ** 2)

class Kite:
    def __init__(self, radius):
        self.radius = radius

rect = Rectangle(2, 3)
circle = Circle(2)

def calculate_area(shape: Shape):
    return shape.area()

print(calculate_area(rect))
print(calculate_area(circle))
# print(calculate_area(Kite(4)))
```
