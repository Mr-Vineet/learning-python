```python
def add(num1: int, num2: int) -> int:
    """adds two numbers

    Args:
        num1 (int): first number
        num2 (int): second number

    Returns:
        int: sum of both the numbers
    """
    num1 *= 2
    return num1 + num2

print(add(3.14159, 4))
print(add(num2=3, num1=4))
```

```python
# Positional Arguments

def participants(*args, age): # function can take any numbers of args
    print(args)
    print(type(args))
    print(age)
    for participant in args:
        print(participant)
        
participants("Vineet", 1, 2,"Pika", age=31)
participants(age=1) #Valid
```

```python
# Keyword Arguments

def person_details(adil, *args, name, **kwargs):
    print(adil)
    print(args)
    print(name)
    for key, value in kwargs.items():
        print(f"{key}: {value}")

person_details(1, 2, 3,"Adil",name="Vineet", age = 21, designation="Dev")
```

```python
# returning multiple values from a fn

def multiply(num1, num2):
    return num1 * num2, num1

print(type(multiply(2, 3)))
```
