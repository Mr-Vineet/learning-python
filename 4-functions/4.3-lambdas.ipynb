### Map and filter


```python
numbers = [1, 2, 3, 4, 5]

def square(num):
    return num ** 2

list(map(square,numbers))
list(map(lambda num: num ** 3, numbers))
```

```python
numbers1 = [1, 2, 3, 4, 5]
numbers2 = [6, 7, 8, 9, 10]

list(map(lambda x, y : x + y, numbers1, numbers2))
```

```python
numbers1 = [1, 2, 3, 4, 5]

squares = list(map(square,numbers1))

print(squares)
```

```python
numbers = ['1', '100', 3]
integers = list(map(int, numbers))

print(integers)
```

```python
words = ["apple", "mango", "banana", "cheeku"]
upper_cased_words = list(map(str.upper, words))

print(upper_cased_words)
```

### Filter


```python
people = [{"name": "Vineet", "age": 21},
          {"name": "Deepanshu", "age": 22},
          {"name": "Adil", "age": 42}]

def is_above_25(person):
    return person["age"] > 25

people_above_25 = list(filter(is_above_25, people))

print(people_above_25)
```
