```python
# problem 1 - Temperature Conversion
from typing import Union

def convert_temperature(temp:float, unit: str) -> Union[float, str]:
    """
    converts the temperature between celsius and fahrenheit

    Args:
        temp (float): value of temperature
        unit (str): unit of temperature

    Returns:
        float or string: converted temperature value or Invalid Input
    """
    if(unit == "C"):
        return 9 / 5 * temp + 32
    
    elif(unit == "F"):
        return 5 / 9 * (temp - 32)
    
    return "Invalid Inputs"

print(convert_temperature(100, "C"))
```
